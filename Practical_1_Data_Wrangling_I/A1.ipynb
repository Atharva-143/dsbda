```python
# DSBDA Practical 1: Data Wrangling I

# 1. Import required Python libraries
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler

# 2. Dataset description
print("Dataset Name: Iris Dataset")
print("Source: Kaggle / UCI Machine Learning Repository")
print("URL: https://www.kaggle.com/datasets/uciml/iris")
print("Description: The Iris dataset contains measurements of iris flowers such as sepal length, sepal width, petal length, petal width, and their species classification.\n")

# 3. Load dataset into pandas DataFrame
df = pd.read_csv('Iris.csv')
# df = pd.read_csv('/home/admin1/Desktop/Iris.csv')

# Display first 5 rows
print("First 5 Rows of Dataset:")
print(df.head())

# Display last 5 rows
print("\nLast 5 Rows of Dataset:")
print(df.tail())

# 4. Data Preprocessing

# Check dimensions of dataset
print("\nShape of Dataset (Rows, Columns):", df.shape)

# Check missing values
print("\nMissing Values in Each Column:")
print(df.isnull().sum())

# Check duplicate rows
print("\nDuplicate Rows:", df.duplicated().sum())

# Remove duplicate rows
df = df.drop_duplicates()
print("Duplicate rows removed successfully")

# Basic statistics
print("\nStatistical Summary:")
print(df.describe())

# Variable descriptions
print("\nVariable Description:")
print("sepal.length  -> Sepal Length (Numeric)")
print("sepal.width   -> Sepal Width (Numeric)")
print("petal.length  -> Petal Length (Numeric)")
print("petal.width   -> Petal Width (Numeric)")
print("variety       -> Type of Iris Flower (Categorical)")

# Check data types
print("\nData Types of Variables:")
print(df.dtypes)

# More detailed info
print("\nDataset Information:")
print(df.info())

# 5. Convert categorical to numeric
df['variety'] = df['variety'].map({'Setosa': 0, 'Versicolor': 1, 'Virginica': 2})

print("\nDataset After Converting Categorical 'variety' into Numeric:")
print(df.head())

# 6. Data Normalization
scaler = MinMaxScaler()
df[['sepal.length', 'sepal.width', 'petal.length', 'petal.width']] = scaler.fit_transform(
    df[['sepal.length', 'sepal.width', 'petal.length', 'petal.width']]
)

print("\nDataset After Normalization:")
print(df.head())

# Final dataset info
print("\nFinal Dataset Information:")
print(df.info())
```
