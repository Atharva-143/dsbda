```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
```

```python
print("Dataset Name : Iris Dataset")
print("Source : Kaggle / UCI Machine Learning Repository")
print("URL : https://www.kaggle.com/datasets/uciml/iris")
```

```python
print("\nDescription :")
print("The Iris dataset contains measurements of iris flowers such as sepal length, sepal width, petal length, petal width and species classification.")
```

```python
df = pd.read_csv("Iris.csv")

print("\nDataset Loaded Successfully")
```

```python
print("\nFirst 5 Rows of Dataset")
print(df.head())
```

```python
print("\nLast 5 Rows of Dataset")
print(df.tail())
```

```python
print("\nShape of Dataset")
print(df.shape)
```

```python
print("\nColumn Names")
print(df.columns)
```

```python
print("\nMissing Values in Each Column")
print(df.isnull().sum())
```

```python
print("\nNumber of Duplicate Rows")
print(df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("\nDuplicate Rows Removed Successfully")
```

```python
print("\nStatistical Summary")
print(df.describe())
```

```python
print("\nDataset Information")
print(df.info())
```

```python
print("\nVariable Description")
print("sepal.length  -> Sepal Length (Numeric)")
print("sepal.width   -> Sepal Width (Numeric)")
print("petal.length  -> Petal Length (Numeric)")
print("petal.width   -> Petal Width (Numeric)")
print("variety       -> Type of Iris Flower (Categorical)")
```

```python
print("\nData Types of Variables")
print(df.dtypes)
```

```python
scaler = MinMaxScaler()

df[['sepal.length',
    'sepal.width',
    'petal.length',
    'petal.width']] = scaler.fit_transform(
    df[['sepal.length',
        'sepal.width',
        'petal.length',
        'petal.width']]
)
```

```python
print("\nDataset After Normalization")
print(df.head())
```

```python
df['variety'] = df['variety'].map({
    'Setosa': 0,
    'Versicolor': 1,
    'Virginica': 2
})

print("\nDataset After Converting Categorical Variable into Numeric")
print(df.head())
```

```python
print("\nFinal Dataset Information")
print(df.info())
```
