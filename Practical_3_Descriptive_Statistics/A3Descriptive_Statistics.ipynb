```python
import pandas as pd
```

```python
data = {
    'Age': [25, 30, 25, 40, 55, 60, 33, 28, 45, 50],
    'Income': [50000, 60000, 52000, 70000, 80000, 90000, 65000, 62000, 75000, 85000],
    'Age_Group': ['20-30', '30-40', '20-30', '40-50', '50-60',
                  '50-60', '30-40', '20-30', '40-50', '50-60']
}
```

```python
df = pd.DataFrame(data)

print("Dataset")
print(df)
```

```python
print("\nSummary Statistics of Income Grouped by Age Group")
print(df.groupby('Age_Group')['Income'].describe())
```

```python
print("\nMean Income")
print(df.groupby('Age_Group')['Income'].mean())
```

```python
print("\nMedian Income")
print(df.groupby('Age_Group')['Income'].median())
```

```python
print("\nMinimum Income")
print(df.groupby('Age_Group')['Income'].min())
```

```python
print("\nMaximum Income")
print(df.groupby('Age_Group')['Income'].max())
```

```python
print("\nStandard Deviation")
print(df.groupby('Age_Group')['Income'].std())
```

```python
iris = pd.read_csv("Iris.csv")
```

```python
print("\nFirst 5 Rows of Iris Dataset")
print(iris.head())
```

```python
print("\nStatistical Details of Iris Dataset Grouped by Species")
print(iris.groupby('variety').describe())
```

```python
print("\nMean Values")
print(iris.groupby('variety').mean())
```

```python
print("\nMedian Values")
print(iris.groupby('variety').median())
```

```python
print("\nStandard Deviation")
print(iris.groupby('variety').std())
```

```python
print("\nPercentiles")
print(iris.groupby('variety').quantile([0.25, 0.50, 0.75]))
```
