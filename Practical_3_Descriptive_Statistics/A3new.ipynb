```python
# DSBDA Practical 3: Descriptive Statistics - Measures of Central Tendency and Variability

import pandas as pd

# Part 1: Summary Statistics grouped by categorical variable

data = {
    'age': [25, 30, 25, 40, 55, 60, 33, 28, 45, 50, 25, 30, 28, 30, 25],
    'income': [50000, 60000, 50000, 70000, 80000, 90000, 65000, 62000, 75000, 85000, 50000, 60000, 62000, 70000, 75000],
    'age_group': ['20-30', '30-40', '20-30', '40-50', '50-60', '50-60', '30-40', '20-30', '40-50', '50-60', '20-30', '30-40', '20-30', '30-40', '20-30']
}

df = pd.DataFrame(data)

print("Dataset:")
print(df)

print("\nSummary Statistics of Income grouped by Age Group:")
summary_stats = df.groupby('age_group')['income'].describe()
print(summary_stats)

print("\nMedian Income by Age Group:")
median_income = df.groupby('age_group')['income'].median()
print(median_income)

print("\nMean Income by Age Group:")
mean_income = df.groupby('age_group')['income'].mean()
print(mean_income)

print("\nMinimum Income by Age Group:")
min_income = df.groupby('age_group')['income'].min()
print(min_income)

print("\nMaximum Income by Age Group:")
max_income = df.groupby('age_group')['income'].max()
print(max_income)

print("\nStandard Deviation of Income by Age Group:")
std_income = df.groupby('age_group')['income'].std()
print(std_income)

print("\nMode of Age:")
print(df['age'].mode().values)

print("Mode of Income:")
print(df['income'].mode().values)

# Part 2: Iris Dataset Statistics

df = pd.read_csv('Iris.csv')

print("\nFirst 5 Rows of Iris Dataset:")
print(df.head())

print("\nSummary Statistics of Iris Dataset grouped by Species:")
summary_stats_species = df.groupby('variety').describe()
print(summary_stats_species)

print("\nPercentiles (25%, 50%, 75%) by Species:")
percentiles = df.groupby('variety').quantile([0.25, 0.50, 0.75])
print(percentiles)

print("\nSummary Statistics by Species for Sepal Width:")
sepal_width_stats = df.groupby('variety')['sepal.width'].describe()
print(sepal_width_stats)

print("\nMedian Values by Species:")
median_values = df.groupby('variety').median()
print(median_values)

print("\nMedian Sepal Length by Species:")
median_sepal_length = df.groupby('variety')['sepal.length'].median()
print(median_sepal_length)

print("\nMode of Sepal Width:")
print(df['sepal.width'].mode().values)
```
