```python
# DSBDA Practical 7: Text Analytics

import nltk
import re
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from sklearn.feature_extraction.text import TfidfVectorizer

# Download required resources
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger_eng')

# Sample document
text = """Hello everyone! I am first name last name. I am a loyal KSKA Git user all the way from Sangamwadi Empire.
I have considerable knowledge about life, Python, C++, Java, Rust, Golang and Blockchain.
For every smart contract, I lose one strand of my hair.
In my free time, which by the way, I barely get, I like to swim."""

print("Original Text:\n", text)

# 1. Sentence Tokenization
sent_tokens = sent_tokenize(text)
print("\nSentence Tokenization:")
print(sent_tokens)

# 2. Word Tokenization
word_tokens = word_tokenize(text)
print("\nWord Tokenization:")
print(word_tokens)

# 3. Remove punctuation
clean_text = re.sub('[^a-zA-Z]', ' ', text)
print("\nText after removing punctuation:")
print(clean_text)

# 4. Stop words removal
stop_words = set(stopwords.words('english'))
tokens = word_tokenize(clean_text.lower())
filtered_words = [word for word in tokens if word not in stop_words]

print("\nTokens after Stop Words Removal:")
print(filtered_words)

# 5. POS Tagging
pos_tags = nltk.pos_tag(filtered_words)
print("\nPOS Tagging:")
print(pos_tags)

# 6. Stemming
ps = PorterStemmer()
stemmed_words = [ps.stem(word) for word in filtered_words]
print("\nStemmed Words:")
print(stemmed_words)

# 7. Lemmatization
lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]
print("\nLemmatized Words:")
print(lemmatized_words)

# 8. TF-IDF Representation
documents = [text]

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(documents)
feature_names = vectorizer.get_feature_names_out()

print("\nTF-IDF Representation:")
print("Feature Names:", feature_names)
print("TF-IDF Matrix:\n", tfidf_matrix.toarray())
```
